```python
import time
import random
```

```python
# Size of dataset
n = 1_000_000

# Maximum random number to generate
N_MAX = 1_000_000_000

# Generate n numbers from 1 to N_MAX
random.seed(123)
data = random.sample(range(N_MAX), n)

assert len(set(data)) == n, 'Duplicates Found!'
```

```python
data[:5]
```

# Linear Search

This function searches for the `target` in our array - it loops through every element to look for it and is thus $O(n)$.

```python
def linear_search(arr, target):
    for value in arr:
        if value == target:
            return True
    return False
```

```python
linear_search(data, random.choice(data))
```

```python
linear_search(data, -5)
```

# Binary Tree

We define a `Node` class that will represent one node of a binary tree. We'll use this in our binary search algorithm next.

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        
    # Recursively add nodes in their correct place
    def add_node(self, new_value):
        if new_value < self.value:
            if not self.left:
                self.left = Node(new_value)
            else:
                self.left.add_node(new_value)
        elif new_value > self.value:
            if not self.right:
                self.right = Node(new_value)
            else:
                self.right.add_node(new_value)
                
    # Recursively search the tree for our target.
    def search(self, target):
        if self.value == target:
            return True
        elif target < self.value:
            return self.left.search(target) if self.left else False
        elif target > self.value:
            return self.right.search(target) if self.right else False
    
    # Make printing of nodes to be pretty
    def __repr__(self):
        return f"Node({self.value})"
```

## Create binary tree from array

Note that this operation itself is $O(n)$!

```python
tree = Node(data[0])

for value in data[1:]:
    tree.add_node(value)    
```

## Binary search function

It's not much, but it kicks off the search.

```python
def binary_search(tree, target):
    return tree.search(target)
```

```python
binary_search(tree, random.choice(data))
```

```python
random.seed(789)
good_target = random.choice(data)
bad_target = -1
```

```python
start = time.time()
binary_search(tree, good_target)
end = time.time()

bst_duration_good = end - start
print(f"Binary Search (found) Took: {bst_duration_good} seconds")

start = time.time()
binary_search(tree, bad_target)
end = time.time()

bst_duration_bad = end - start
print(f"Binary Search (not found) Took: {bst_duration_bad} seconds")
```

```python
start = time.time()
linear_search(data, good_target)
end = time.time()

lin_duration_good = end - start
print(f"Linear Search (found) Took: {lin_duration_good} seconds")

start = time.time()
linear_search(data, bad_target)
end = time.time()

lin_duration_bad = end - start
print(f"Linear Search (not found) Took: {lin_duration_bad} seconds")
```

## Comparison

```python
print(f"When finding the target, binary search is {round(lin_duration_good / bst_duration_good, 2)} times faster than linear search.")
print(f"When NOT finding the target, binary search is {round(lin_duration_bad / bst_duration_bad, 2)} times faster than linear search.")
```
