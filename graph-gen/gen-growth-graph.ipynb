```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
xs = np.linspace(1, 10, 10_000)
xslog = xs[xs >= 2]
```

```python
funcs = {
    '$\log n$': lambda n: np.log2(n),
    '$n$': lambda n: n,
    '$n \log n$': lambda n: n*np.log2(n),
    '$n^2$': lambda n: n**2,
    '$2^n$': lambda n: 2**n
}
```

```python
plt.figure(figsize=(10, 6))

plt.plot(xs, xs**xs, color='pink', label='$n^n$', linewidth=2)
plt.plot(np.arange(1, 11), [math.factorial(n) for n in np.arange(1, 11)], color='green', label='$n!$', linewidth=2)
plt.plot(xs, 2**xs, color='purple', label='$2^n$', linewidth=2)
plt.plot(xs, xs**2, color='navy', label='$n^2$', linewidth=2)
plt.plot(xslog, xslog*np.log2(xslog), color='skyblue', label='$n \log n$', linewidth=2)
plt.plot(xs, xs, color='red', label='$n$', linewidth=2)
plt.plot(xslog, np.log2(xslog), color='orange', label='$\log n$', linewidth=2)
    
plt.legend()
plt.yscale('log')

plt.tight_layout()
plt.savefig('big-o-growth.png', dpi=192)
```
