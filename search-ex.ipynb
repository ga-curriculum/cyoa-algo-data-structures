```python
import random
import time
```

```python
n = 1_000_000

N_MAX = 1_000_000_000
random.seed(123)
data = random.sample(range(N_MAX), n)

assert len(set(data)) == n, 'Duplicates Found!'
```

```python
data[:5]
```

```python
def linear_search(arr, target):
    for value in arr:
        if value == target:
            return True
    return False
```

```python
%%time
linear_search(data, random.choice(data))
```

```python
%%time
linear_search(data, -5)
```

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        
    def add_node(self, new_value):
        if new_value < self.value:
            if not self.left:
                self.left = Node(new_value)
            else:
                self.left.add_node(new_value)
        elif new_value > self.value:
            if not self.right:
                self.right = Node(new_value)
            else:
                self.right.add_node(new_value)
                
    def search(self, target):
        if self.value == target:
            return True
        elif target < self.value:
            return self.left.search(target) if self.left else False
        elif target > self.value:
            return self.right.search(target) if self.right else False
    
    def __repr__(self):
        return f"Node({self.value})"
```

```python
root = Node(data[0])

for value in data[1:]:
    root.add_node(value)    
```

```python
root
```

```python
def binary_search(tree, target):
    return tree.search(target)
```

```python
binary_search(root, data[0])
```

```python
%%time
binary_search(root, random.choice(data))
```

```python
# %%timeit
start = time.time()
binary_search(root, data[-1])
end = time.time()
print(end - start)
```

```python
# %%timeit
start = time.time()
linear_search(data, data[-1])
end = time.time()
print(end - start)
```

```python
random.seed(456)
my_list = [random.randint(1, 101) for _ in range(15)]
```

```python
my_list
```

```python
sorted_list = []
previous_min = -1

for i in range(len(my_list)):
    current_min = 999
    for item in my_list:
        if item < current_min and item > previous_min:
            current_min = item
    sorted_list.append(current_min)
    previous_min = current_min
```
